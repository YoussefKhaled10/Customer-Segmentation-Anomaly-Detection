### Clean StockCode by removing trailing letters and check unique values again

```python
df["StockCode_clean"] = df["StockCode"].astype(str).str.replace(
    r"[A-Za-z]$", 
    "", 
    regex=True
)
df["StockCode_clean"].nunique()
```

```python
df.info()
```

```python
(
    df.groupby("StockCode_clean")["StockCode"]
      .nunique()
      .sort_values(ascending=False)
      .head(10)
)
```

```python
df.loc[
    ~df["StockCode_clean"].astype(str).str.isnumeric(),
    "StockCode_clean"
].unique()
```

### Remove invalid stock codes

```python
invalid_stock_codes = df.loc[
    ~df["StockCode_clean"].astype(str).str.isnumeric(),
    "StockCode_clean"
].unique()
df = df.loc[~df["StockCode_clean"].isin(invalid_stock_codes)]
df["StockCode_clean"] = df["StockCode_clean"].astype(int)
df.info()
```

```python
df.drop(columns=["StockCode"], inplace=True)
```

```python
df.info()
```

```python
df.describe()
```

### Remove rows with negative Quantity or Price

```python
df = df[(df["Quantity"] > 0) & (df["Price"] > 0)]
```

```python
df.describe()
```

### Check Remove or no

```python
df[
    (df["Quantity"] <= 0) |
    (df["Price"] <= 0)
].shape
```

### Make some Feature Engineering

```python
df["TotalPrice"] = df["Quantity"] * df["Price"]
```

```python
df
```

```python
invoice_basket = (
    df.groupby("Invoice")["StockCode_clean"]
      .nunique()
      .reset_index(name="BasketSize")
)
invoice_basket
```

```python
reference_date = df["InvoiceDate"].max() + pd.Timedelta(days=1)
```

```python
rfm = df.groupby("Customer ID").agg({
    "InvoiceDate": lambda x: (reference_date - x.max()).days,  
    "Invoice": "nunique",                                       
    "TotalPrice": "sum"                                         
}).reset_index()

rfm.columns = ["CustomerID", "Recency", "Frequency", "Monetary"]
```

```python
rfm["AvgOrderValue"] = rfm["Monetary"] / rfm["Frequency"]
```
